```python
import re
import unicodedata
normalisasi_kata = {
    'yg': 'yang',
    'utk': 'untuk',
    'dgn': 'dengan',
    'sdh': 'sudah',
    'krn': 'karena',
    'apik': 'bagus',
    'tdk': 'tidak',
    'klo': 'kalo',
    'sbg': 'sebagai',
    'gue': 'aku',
    'dlm': 'dalam',
    'jgn': 'jangan',
    'jkw': 'jokowi',
    'org': 'orang',
    'amp': '',
}
def clean_twitter_text(text):
    # Normalize unicode (hilangkan tanda aksen dsb)
    text = unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode('utf-8', 'ignore')

    # Lowercase
    text = text.lower()
    
    # Hapus URL
    text = re.sub(r"http\S+|www\S+|https\S+", '', text)

    # Hapus mention (@username)
    text = re.sub(r'@\w+', '', text)
    
    # Hapus hashtag symbol, tapi simpan kata (misal: #bahagia -> bahagia)
    text = re.sub(r'#', '', text)
    
    # Hapus emoji dan simbol non-ASCII
    text = re.sub(r'[^\x00-\x7F]+', ' ', text)

    # Hapus angka jika tidak perlu (opsional)
    text = re.sub(r'\d+', '', text)

    # Hapus tanda baca (opsional, bisa simpan tanda tanya/seru kalau mau)
    text = re.sub(r'[^\w\s.,!?]', '', text)

    # Normalize 
    words = text.split()
    words = [normalisasi_kata.get(w, w) for w in words]  
    text = ' '.join(words)

    # Hapus spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text
```

```python
import pandas as pd

pd.set_option('display.max_colwidth', None)  # Tampilkan seluruh isi kolom
```

```python
df['text_processed'] = df['full_text'].apply(clean_twitter_text)
df.head()
```

```python
from collections import Counter
import re

df_copy = df.copy()

# Gabungkan semua teks jadi satu string
text = ' '.join(df_copy['text_processed'].astype(str))

# Preprocessing: lowercase + tokenisasi
words = re.findall(r'\b\w+\b', text.lower())

# Hitung frekuensi kata
freq = Counter(words)

# Ambil 50 kata paling sering
top_50 = freq.most_common(150)

# Cetak hasil
for word, count in top_50:
    print(f"{word}: {count}")
```

```python
import pandas as pd
from googletrans import Translator
from tqdm import tqdm
import time

# Inisialisasi penerjemah
translator = Translator()

# Muat dataset
df = pd.read_csv("processed.csv")  # ganti sesuai nama file kamu
df = df[["text_processed"]].dropna().copy()
df = df.head(500)  # Ambil 500 data pertama
df["translate_en"] = ""

# Terjemahkan baris per baris dengan progress bar
for i in tqdm(df.index, desc="Menerjemahkan Tweet"):
    try:
        teks = df.at[i, "text_processed"]
        hasil = translator.translate(teks, src='id', dest='en')
        df.at[i, "translate_en"] = hasil.text
        time.sleep(0.3)  # delay ringan agar tidak diblokir
    except Exception as e:
        print(f"❌ Error pada baris {i}: {e}")
        df.at[i, "translate_en"] = "error"

# Simpan hasil
df.to_csv("translated_google.csv", index=False, encoding="utf-8-sig")
print("\n✅ Translasi selesai! Hasil disimpan di 'translated_google.csv'")
```
