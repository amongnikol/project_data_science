```python
import pandas as pd
import numpy as np

import mlflow
from mlflow.models import infer_signature

from sklearn.preprocessing import OneHotEncoder,PolynomialFeatures,StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import VotingClassifier,GradientBoostingClassifier,RandomForestClassifier
from sklearn.pipeline import Pipeline as pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler
import os
```

```python
train_data = pd.read_csv('dataset/train.csv')
test_data = pd.read_csv('dataset/test.csv')
```

```python
to_convert_num = train_data.select_dtypes(['int64', 'float64'])
x0 = to_convert_num.drop(columns='is_claim')
y0 = to_convert_num['is_claim']
```

```python
smote = SMOTE()
rus = RandomUnderSampler()
x1, y1 = smote.fit_resample(x0, y0)
x, y = rus.fit_resample(x1, y1)

y.value_counts()
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python

to_convert_nums = to_convert_num.drop(columns='is_claim').columns
preprocess = ColumnTransformer(transformers=[('num',StandardScaler(),to_convert_nums)])
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import RandomizedSearchCV


model_pipeline = pipeline(steps=[
    ('preprocessing', preprocess),
    ('poly_features', PolynomialFeatures()), 
    ('classifier', SVC(random_state=42)) 
])

params = {
    'poly_features__degree': [1, 2, 3], # Derajat polinomial
    'classifier__C': np.logspace(-3, 2, 6), # Parameter regularisasi C untuk SVC
    'classifier__kernel': ['linear', 'rbf', 'poly', 'sigmoid'], # Kernel untuk SVC
    'classifier__gamma': ['scale', 'auto'] + list(np.logspace(-3, 2, 6)), # Gamma untuk kernel rbf, poly, sigmoid
    'classifier__probability': [True] # Diperlukan jika Anda ingin mendapatkan probabilitas
}

model = RandomizedSearchCV(
    estimator=model_pipeline,
    param_distributions=params,
    n_iter=20, 
    cv=3, 
    scoring='accuracy', 
    verbose=2, 
    random_state=42,
    n_jobs=-1 
)

model.fit(x_train, y_train)
print(model.best_params_)
print(model.score(x_train, y_train), model.best_score_, model.score(x_test, y_test))
```

```python
score = model.predict(x_test)
accuracy = accuracy_score(y_test, score)
accuracy
```
