### Importing Libraries & Loading Data

```python
# Import necessary libraries for data analysis and manipulation
import pandas as pd
import numpy as np

# Import libraries for data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Import machine learning libraries and modules
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error
import xgboost as xgb

# Configure notebook for better visualization and output
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (14, 7)
pd.set_option('display.max_columns', 50)

# Load the datasets
train_df = pd.read_csv('Training_set_augmented.csv')
test_df = pd.read_csv('Public_Test_Set.csv')
sample_submission = pd.DataFrame({'ID': test_df['ID']})

# --- Critical Best Practice: Clean column names ---
# This removes any accidental leading/trailing spaces from column names
train_df.columns = train_df.columns.str.strip()
test_df.columns = test_df.columns.str.strip()

print("Datasets loaded and column names cleaned successfully!")
print(f"Training data shape: {train_df.shape}")
print(f"Test data shape: {test_df.shape}")
```

### Feature Engineering

Di sinilah kita bisa menjadi kreatif dan menambahkan nilai yang signifikan. Kita akan membuat fitur-fitur baru yang menangkap informasi yang lebih mendalam dibandingkan hanya menggunakan kolom asli. Fitur-fitur baru yang akan kita buat meliputi:

- GDP_per_Capita: Ukuran dari output ekonomi per orang.

- Production_Efficiency: Seberapa banyak energi yang diproduksi dibandingkan dengan kapasitas terpasang? Ini menunjukkan efisiensi operasional.

- Lag_Production_1_Year: Produksi dari jenis energi tertentu di suatu negara pada tahun sebelumnya. Ini sering menjadi prediktor yang sangat kuat.

```python
import numpy as np
import pandas as pd

def feature_engineer(df):
    """Engineers new features for the model."""
    df_copy = df.copy()

    # Sort data to correctly calculate lag features
    df_copy = df_copy.sort_values(by=['Country', 'Energy Type', 'Year']).reset_index(drop=True)

    # Check if 'GDP (USD)' and 'Population' columns exist
    if 'GDP (USD)' in df_copy.columns and 'Population' in df_copy.columns:
        # Economic Features
        df_copy['GDP_per_Capita'] = df_copy['GDP (USD)'] / df_copy['Population']
    else:
        df_copy['GDP_per_Capita'] = np.nan

    # Efficiency Features
    # Replace 0 in capacity to avoid division by zero
    if 'Production (GWh)' in df_copy.columns and 'Installed Capacity (MW)' in df_copy.columns:
        df_copy['Production_Efficiency'] = df_copy['Production (GWh)'] / df_copy['Installed Capacity (MW)'].replace(0, np.nan)
    else:
        df_copy['Production_Efficiency'] = np.nan

        # Lag Features
    # This captures the production from the previous year for the same country and energy type
    if 'Production (GWh)' in df_copy.columns:
        df_copy['Lag_Production_1_Year'] = df_copy.groupby(['Country', 'Energy Type'])['Production (GWh)'].shift(1)
    else:
        df_copy['Lag_Production_1_Year'] = np.nan

    # Fill any resulting NaNs from feature engineering, perhaps with a median or 0
    df_copy.fillna(0, inplace=True)  # Simple fill for demonstration, more complex imputation happens in pipeline

    return df_copy

# Apply feature engineering to a copy of the original dataframes
train_featured = feature_engineer(train_df.copy())
test_featured = feature_engineer(test_df.copy())

# Align columns - crucial for consistent preprocessing
if 'Production (GWh)' in train_featured.columns:
    train_labels = np.log1p(train_featured['Production (GWh)'])
else:
    train_labels = np.nan

train_ids = train_featured['ID'] if 'ID' in train_featured.columns else train_featured.index

# Drop target and non-feature columns
if 'Production (GWh)' in train_featured.columns:
    train_featured = train_featured.drop(columns=['Production (GWh)'])
if 'ID' in train_featured.columns:
    train_featured = train_featured.drop(columns=['ID'])

# Ensure test set has the same columns as the training set (except the target)
# and in the same order
train_cols = train_featured.columns
test_featured = test_featured[train_cols]

print("Feature engineering complete.")
print(f"New training data shape: {train_featured.shape}")
```

### Modeling

```python
# Identify categorical and numerical features
categorical_features = train_featured.select_dtypes(include=['object', 'category']).columns
numerical_features = train_featured.select_dtypes(include=np.number).columns

# Preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ('num', Pipeline(steps=[
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())]), numerical_features),
        ('cat', Pipeline(steps=[
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('onehot', OneHotEncoder(handle_unknown='ignore'))]), categorical_features)
    ],
    remainder='passthrough'
)

# Cross-Validation setup
NFOLDS = 5
folds = KFold(n_splits=NFOLDS, shuffle=True, random_state=42)
oof_preds = np.zeros(train_featured.shape[0])
sub_preds = np.zeros(test_featured.shape[0])

print("🚀 Starting Cross-Validation Training with Enhanced Reporting...\n")

# Loop over each fold
for n_fold, (train_idx, valid_idx) in enumerate(folds.split(train_featured, train_labels)):
    X_train, y_train = train_featured.iloc[train_idx], train_labels.iloc[train_idx]
    X_valid, y_valid = train_featured.iloc[valid_idx], train_labels.iloc[valid_idx]

    # Model definition
    lgb_model = lgb.LGBMRegressor(
        random_state=42,
        n_estimators=1000,
        learning_rate=0.05,
        num_leaves=31,
        verbosity=-1  # suppress LightGBM output
    )

    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('regressor', lgb_model)
    ])

    # Fit model
    pipeline.fit(X_train, y_train)

    # Predictions
    val_preds = pipeline.predict(X_valid)
    train_preds = pipeline.predict(X_train)

    oof_preds[valid_idx] = val_preds
    sub_preds += pipeline.predict(test_featured) / NFOLDS

    # Metrics
    val_rmse = np.sqrt(mean_squared_error(y_valid, val_preds))
    train_rmse = np.sqrt(mean_squared_error(y_train, train_preds))
    val_perc_error = (val_rmse / np.mean(y_valid)) * 100 if np.mean(y_valid) != 0 else 0
    train_perc_error = (train_rmse / np.mean(y_train)) * 100 if np.mean(y_train) != 0 else 0

    # Reporting
    print(f"================= FOLD {n_fold+1}/{NFOLDS} =================")
    print(f"🎯 Validation RMSE : {val_rmse:.4f} (Error: {val_perc_error:.2f}%)")
    print(f"📉 Training RMSE   : {train_rmse:.4f} (Error: {train_perc_error:.2f}%)")
    print("=" * 43)

# Final summary
overall_rmse = np.sqrt(mean_squared_error(train_labels, oof_preds))
overall_percentage_rmse = (overall_rmse / np.mean(train_labels)) * 100 if np.mean(train_labels) != 0 else 0

print("\n===============================")
print("🏆 OVERALL CV SUMMARY 🏆")
print("================================")
print(f"✅ Average Validation RMSE: {overall_rmse:.4f}")
print(f"📊 Average Validation Error: {overall_percentage_rmse:.2f}%")


```

### Final Prediction

```python
# Inverse transform the predictions to get them back to the original scale
predictions = np.expm1(sub_preds)

# It's good practice to ensure predictions are non-negative
predictions[predictions < 0] = 0

# Create the submission DataFrame
submission_df = pd.DataFrame({'ID': sample_submission['ID'], 'Predicted Production (GWh)': predictions})

# Save the submission file
submission_df.to_csv('submission.csv', index=False)

print("\n'submission.csv' has been created successfully!")
print("The file is ready for submission. Good luck!")
print("\nSubmission File Head:")
print(submission_df.head())
```
